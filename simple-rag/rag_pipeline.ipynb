```python
from langchain_groq import ChatGroq
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface.embeddings import HuggingFaceEmbeddings
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
llm = ChatGroq(model=os.getenv('GROQ_MODEL'))
embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
loader = PyPDFLoader('AI_Engineering_Roadmap_2026.pdf')

data = loader.load()

text_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=100)

documents = text_splitter.split_documents(data)
```

```python
vector_store = InMemoryVectorStore.from_documents(documents=documents, embedding=embedding)

while True:
    query = input("User: ")
    if query.lower() == "exit":
        break

    context = ""
    chunks = vector_store.similarity_search(query)
    for chunk in chunks:
        context += f"{chunk.page_content}\n\n"

    res = llm.invoke(f'''You are an AI Assistant who can give answers of user query only according to the context provided
        Query: {query}
        Context: {context}
    ''')

    print(res.content)
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
template = PromptTemplate.from_template("""
        You are an AI Assistant who can give answers of user query only according to the context provided
        Query: {query}
        Context: {context}
""")

chain = template | llm | StrOutputParser()

while True:
    query = input("User: ")
    if query.lower() == "exit":
        break

    context = ""
    chunks = vector_store.similarity_search(query)
    for chunk in chunks:
        context += f"{chunk.page_content}\n\n"

    res = chain.invoke({'query':query, 'context':context})

    print(res)
```
