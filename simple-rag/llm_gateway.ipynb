```python
from litellm import completion
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
groq_response = completion(
    model=f'groq/{os.getenv('GROQ_MODEL')}',
    messages=[{'role':'user', 'content':'what is rag'}]
)

print(groq_response.choices[0].message.content)
```

```python
google_response = completion(
    model=f"gemini/{os.getenv('GOOGLE_MODEL')}",
    api_key=os.getenv('GOOGLE_API_KEY'),
    messages=[{'role':'user', 'content':'hi'}]
)

print(google_response.choices[0].message.content)
```
