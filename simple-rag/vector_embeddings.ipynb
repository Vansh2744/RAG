```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embeddings = HuggingFaceEmbeddings(model="sentence-transformers/all-MiniLM-L6-v2")
```

```python
query = "What is the method to call a user"

vector_embeddings = embeddings.embed_query(query)

len(vector_embeddings)
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.vectorstores import InMemoryVectorStore
```

```python
loader = PyPDFLoader('AI_Engineering_Roadmap_2026.pdf')

data = loader.load()

vector_store = InMemoryVectorStore.from_documents(documents=data, embedding=embeddings)
```

```python
chunks = vector_store.similarity_search(query="AI Interview", k=2)

for chunk in chunks:
    print(chunk.page_content)
```
